```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import colormaps as cmap
```

```python
L=14
N=L**3
T_min, T_max = 0.2, 2.0
NB=48
T_list = np.linspace(T_min, T_max, NB)
Tc = 0.95
labels = [
    "first of majority" ,
    "first of minority" ,
    "second of majority",
    "second of minority",
    "third of majority" ,
    "third of minority" ,
]

ctype1 = "winter"
ctype2 = "autumn"

n = 9

colors1 = cmap[ctype1](np.linspace(0.16, 0.85, n))
colors2 = cmap[ctype2](np.linspace(0., 0.75, n))

fig, ax = plt.subplots(figsize=(6, 5))

for j in range(n):
    T_list, y1, y2 = np.loadtxt(f"DATA/clustats_first_{j}.txt")
    ax.plot(T_list, y1, label=labels[2], c=colors1[j])
    ax.plot(T_list, y2, label=labels[3], c=colors2[j])

ax.axhline(N // 2, ls="--", alpha=0.5, c="gray")
ax.axvline(Tc, ls="--", alpha=0.5, c="gray")

ax.set_xlim(0.2, 1.5)
ax.set_ylim(0, N)
ax.set_xlabel(r"$T$")
ax.set_ylabel("Cluster size")

fig.tight_layout()
#plt.savefig("FIG1-l.pdf")
plt.show()
```

```python
fig, axes = plt.subplots(3, 3, figsize=(6, 5), sharex=True, sharey=True)
axes = axes.ravel()
j = 0
for j, ax in enumerate(axes):

    data = np.load(f"DATA/clustats_others_{j}.npy")
    for i in range(4):
        ax.plot(T_list, data[i], label=labels[i+2], c=f"C0{i+2}")
    ax.axhline(N/2, ls="--", alpha=0.5, c="gray")
    ax.axvline(Tc, ls="--", alpha=0.5, c="gray")


ax.set_xlim(0.2, 1.5)
#ax.set_yscale("log")
ax.set_ylim(0, 57)

# Optional: only put labels on outer axes to avoid clutter
for ax in axes[-3:]:
    ax.set_xlabel(r"$T$")
for ax in axes[::3]:
    ax.set_ylabel("Cluster size")

# One shared legend for the whole figure
handles, legend_labels = axes[0].get_legend_handles_labels()
#fig.legend(handles, legend_labels)#, loc="upper right")
fig.subplots_adjust(wspace=0.05, 
                    hspace=0.05)
#plt.savefig("FIG1-r.pdf")
#fig.tight_layout()
plt.show()
```

```python
import numpy as np
x = np.loadtxt("DATA/chaosData/temps.dat")
hardcaos = np.loadtxt("DATA/chaosData/hard.dat")
allcaos = np.loadtxt("DATA/chaosData/all.dat")
allcaos_avg, allcaos_err = np.mean(allcaos, axis = 0), np.std(allcaos, axis = 0)/np.sqrt(50-1)
argmax = np.argmax(allcaos_avg-hardcaos[-1])
highestdiff = allcaos.T[argmax]
sorted_caos = np.argsort(highestdiff)
perc16, perc84 = allcaos[sorted_caos[8]], allcaos[sorted_caos[42]]
mediancaos = allcaos[sorted_caos[25]]
from matplotlib import colormaps as color

ctype, cidx = "winter", 0.5
c1 = color[ctype](cidx)
cmap = color["hot"]   # choose any colormap you like
colors = cmap(np.linspace(0.5, 0.1, len(hardcaos)))

import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize = (12, 5))


########################################################
tau2 = np.loadtxt("DATA/tau2ALL_L14.txt")
tau5 = np.loadtxt("DATA/tau5HARD_L14.txt")

bins = np.linspace(tau2.min() - 15, tau2.max() + 15, 40)
ax[0].hist(tau2, bins=bins, label=r'All samples', color=c1)
ax[0].hist(tau5, bins=bins, histtype="step", linewidth=2, label=r'Hardest', color=colors[-1])
ax[0].set_xlabel(r'$\tau_\mathrm{TRW}$')
ax[0].set_ylabel(r'Count')
ax[0].legend()

########################################################

ctype, cidx = "winter", 0.5
c1 = color[ctype](cidx)
cmap = color["hot"]   # choose any colormap you like
colors = cmap(np.linspace(0.5, 0.1, len(hardcaos)))

for y, c in zip(hardcaos[:-1], colors[:-1]):
    ax[1].plot(x[:-1], y, color=c)
ax[1].plot(x[:-1], hardcaos[-1], color=colors[-1], label="Hardest")
#ax[1].plot(x[:-1], allcaos_avg, color=c1, label="Typical")
ax[1].plot(x[:-1], perc16, "--", c=c1)
ax[1].plot(x[:-1], perc84, "--", c=c1)
ax[1].plot(x[:-1], mediancaos, c=c1, label="Typical")

textparams = {"size":9.5}
ax[1].text(0.54, 0.29, "bin 0", fontdict=textparams)
ax[1].text(0.79, 0.505, "bin 4", fontdict=textparams)
ax[1].annotate("", xytext=(0.95, 0.5), xy=(0.66, 0.26),
            arrowprops=dict(arrowstyle="->"))

ax[1].set_xlabel("$T$")
ax[1].set_ylabel(r"$\bar{X}(T, T_\mathrm{min})$")
ax[1].grid()
ax[1].legend()

########################################################

plt.tight_layout()
#plt.savefig("FIG5.pdf")
plt.show()
plt.close()
```
